## Parte 1

El cifrado Vigenère es un metodo simple que se usa para ocultar mensajes usando una palabra clave la cual cambia las letras del texto original de forma repetida. Siendo un crifrado facil de entender, ya que este puede ser aplicable a mano.

Ejemplo de aplicacion
Imagina que quieres cifrar el mensaje "HOLA AMIGO" con la clave "SOL". ya que la clave es mas corta que el mensaje se repite hasta cubrir todo el mensaje "SOLSOLSOL".
Luego para cada letra se suma el numero de la posicion de la clave (S=18), (O=14), (L=11) a la del mensaje (A=0), (B=1), (C=2), ..., (Z=25) y tomas el resto al dividirlo por 26
H (7) + S (18) = 25 → Z
O (14) + O (14) = 28 mod 26 = 2 → C
L (11) + L (11) = 22 → W
A (0) + S (18) = 18 → S

Y así sigue: el resultado es "ZCWSE UZNVG". 



¿Por que lo elegi?
En un ejercicio en clase hicimos uno cifrado similar sin saber de que se trataba de este, es por eso que decidi investigar bien como es que funcionaba


Ventajas
- Es mas seguro que los metodos simples: Cambia las letras con varias llaves siendo polialfabetico, lo que confunde el analisis de letras repetidas como la "E" en el español, ya que es la letra mas usada
- Es facil de usar, ya que solo se necesita realizar una matriz 
- Es flexible porque se puede usar palabra como clave y ajustar al alfabeto 


Vulnerabilidades
- Si la clave es mucho mas corta que la palabra se puede cifrar con facilidad, ya que se puede usar Kasiski el cual busca palabras repetidas
- Si el mensaje es largo se pueden contar las letras para contar la clave
- Las computadoras rompen sumamente rapido el cifrado

##### Realizar la creación de un script que permita la conversión de palabras en texto ASCII a BINARIO

```python
# Datos de prueba
# ASCII: hola
# Binario: 01101000011011110110110001100001
# Base64: aG9sYQ==


#XOR
# 01001101
# 00110011
```

```python
def ascii_a_binario(texto):
    resultado = []
    
    # Por cada caracter
    for caracter in texto:
        # Obtenemos su numero ASCII
        ascii_val = ord(caracter)  

        binario = ''
        numero = ascii_val 

        while numero > 0:
            residuo = numero % 2
            binario = str(residuo) + binario
            numero = numero // 2

        while len(binario) < 8:
            binario = '0' + binario

        # Lo agregamos a una lista 
        resultado.append(binario)
    
    return resultado


palabra = input("Ingrese una palabra o frase: ")

binarios = ascii_a_binario(palabra)

print("\nTexto:")
print(palabra)

print("\nBinario:")
binario_total = ""
for b in binarios:
    binario_total += b
print(binario_total)

for i in range(len(palabra)):
    print(palabra[i], "->", binarios[i])

```

##### Realizar la creación de un script que permita la conversión de palabras en texto BASE64 a BINARIO

```python
alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789+/"

# Convierte un byte a binario de 8 bits
def byte_a_bin(valor):
    binario = ""
    numero = valor

    if numero == 0:
        binario = "0"

    while numero > 0:
        residuo = numero % 2
        binario = str(residuo) + binario
        numero = numero // 2

    while len(binario) < 8:
        binario = "0" + binario

    return binario


# Convierte Base64 a lista de bytes
def base64_a_bytes(b64):
    limpio = ""

    for c in b64:
        if c != " " and c != "\n" and c != "\t" and c != "\r":
            limpio += c

    if len(limpio) % 4 != 0:
        print("Error: Base64 invalido")
        return []

    Bytes = []
    i = 0

    while i < len(limpio):
        bloque = limpio[i:i+4]

        pad = 0
        if bloque[3] == "=":
            pad += 1
        if bloque[2] == "=":
            pad += 1

        nums = []

        for c in bloque:
            if c == "=":
                nums.append(0)
            else:
                pos = alfabeto.find(c)
                if pos == -1:
                    print("Error: caracter invalido:", c)
                    return []
                nums.append(pos)

        a = nums[0]
        b = nums[1]
        c = nums[2]
        d = nums[3]

        total = 0
        total += a * (2 ** 18)
        total += b * (2 ** 12)
        total += c * (2 ** 6)
        total += d

        byte1 = (total // (2 ** 16)) % 256
        byte2 = (total // (2 ** 8)) % 256
        byte3 = total % 256

        if pad == 0:
            Bytes.append(byte1)
            Bytes.append(byte2)
            Bytes.append(byte3)
        elif pad == 1:
            Bytes.append(byte1)
            Bytes.append(byte2)
        elif pad == 2:
            Bytes.append(byte1)

        i += 4

    return Bytes


# Convierte Base64 a binario
def base64_a_bin(b64):
    Bytes = base64_a_bytes(b64)
    binario = ""

    for b in Bytes:
        binario += byte_a_bin(b)

    return binario


texto = input("Ingrese texto en Base64: ")

print("\nBase64:")
print(texto)

binario = base64_a_bin(texto)

print("\nBinario:")
print(binario)
```

##### Realizar la creación de un script que permita la conversión de BINARIO a BASE64

```python
Alfanumerico = "ABCDEFGHIJKLMNOPQRSTUVWXYZabcdefghijklmnopqrstuvwxyz0123456789+/"

# Convierte 8 bits a un byte (0-255)
def bits_a_byte(bits):
    valor = 0
    pot = 0

    i = len(bits) - 1
    while i >= 0:
        if bits[i] == "1":
            valor += 2 ** pot
        pot += 1
        i -= 1

    return valor


# Convierte binario a lista de Bytes
def bin_a_bytes(bin_txt):
    limpio = ""

    for c in bin_txt:
        if c != " " and c != "\n" and c != "\t" and c != "\r":
            limpio += c

    if len(limpio) % 8 != 0:
        print("Error: el binario debe ser multiplo de 8")
        return []

    Bytes = []
    i = 0

    while i < len(limpio):
        bits8 = limpio[i:i+8]
        byte = bits_a_byte(bits8)
        Bytes.append(byte)
        i += 8

    return Bytes


# Convierte Bytes a Base64 manualmente
def bytes_a_base64(Bytes):
    res = ""
    i = 0

    while i < len(Bytes):
        b1 = Bytes[i]
        b2 = 0
        b3 = 0

        faltan = len(Bytes) - i
        if faltan >= 2:
            b2 = Bytes[i+1]
        if faltan >= 3:
            b3 = Bytes[i+2]

        total = 0
        total += b1 * (2 ** 16)
        total += b2 * (2 ** 8)
        total += b3

        s1 = (total // (2 ** 18)) % 64
        s2 = (total // (2 ** 12)) % 64
        s3 = (total // (2 ** 6)) % 64
        s4 = total % 64

        c1 = Alfanumerico[s1]
        c2 = Alfanumerico[s2]
        c3 = Alfanumerico[s3]
        c4 = Alfanumerico[s4]

        if faltan == 1:
            c3 = "="
            c4 = "="
        if faltan == 2:
            c4 = "="

        res += c1
        res += c2
        res += c3
        res += c4

        i += 3

    return res


bin_ing = input("Ingrese un binario: ")

print("\nBinario ingresado:")
print(bin_ing)

Bytes = bin_a_bytes(bin_ing)
base64 = bytes_a_base64(Bytes)

print("\nBase64:")
print(base64)

```

##### Realizar la creación de un script que permita la conversión de BINARIO a ASCII

```python
# Convierte 8 bits a un numero ASCII
def bits_a_ascii(bits):
    valor = 0
    potencia = 0

    i = len(bits) - 1
    while i >= 0:
        if bits[i] == "1":
            valor += 2 ** potencia
        potencia += 1
        i -= 1

    return chr(valor)


# Convierte binario a texto ASCII
def binario_a_ascii(binario):
    limpio = ""

    # Limpia espacios y saltos de linea
    for c in binario:
        if c != " " and c != "\n" and c != "\t" and c != "\r":
            limpio += c

    texto = ""
    i = 0

    while i < len(limpio):
        bits8 = limpio[i:i+8]
        letra = bits_a_ascii(bits8)
        texto += letra
        i += 8

    return texto


bin_ing = input("Ingrese el binario: ")

texto = binario_a_ascii(bin_ing)

print("\nBinario:")
print(bin_ing)

print("\nTexto:")
print(texto)
```

##### Realizar la creación de un script que permita aplicar XOR a un BINARIO 

```python
# Aplica XOR entre dos binarios del mismo largo
def xor_binario(bin1, bin2):
    res = ""
    i = 0

    while i < len(bin1):
        b1 = bin1[i]
        b2 = bin2[i]

        if b1 == b2:
            res += "0"
        else:
            res += "1"

        i += 1

    return res


bin_a = input("Ingrese el primer binario: ")
bin_b = input("Ingrese el segundo binario: ")

# Limpiar espacios
a = ""
b = ""

for c in bin_a:
    if c == "0" or c == "1":
        a += c

for c in bin_b:
    if c == "0" or c == "1":
        b += c

if len(a) != len(b):
    print("\nError: los binarios deben tener el mismo largo")
else:
    resultado = xor_binario(a, b)

    print("\nBinario A:")
    print(a)

    print("\nBinario B:")
    print(b)

    print("\nResultado XOR:")
    print(resultado)
```

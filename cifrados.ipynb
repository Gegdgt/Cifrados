### Reutilizando el codigo del ejercicio de historicos:

```python
abc_may = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
abc_min = "abcdefghijklmnopqrstuvwxyz"
```

```python
import string

def cesar_descifrar(msj, d):
    # Diccionarios de sustitucion inversa
    may = {}
    min = {}

    d = d % 26

    # construir diccionarios inversos
    for i in range(26):
        may[abc_may[(i + d) % 26]] = abc_may[i]
        min[abc_min[(i + d) % 26]] = abc_min[i]

    res = ""

    # descifrar caracter por caracter
    for ch in msj:
        if ch in may:
            res += may[ch]
        elif ch in min:
            res += min[ch]
        else:
            res += ch

    return res


# Función ROT13
def ROT13_descifrar(texto):
    res = cesar_descifrar(texto, 13)
    
    return res


# Mensajes
mensaje_cesar = "IODJ{FHVDU_FLIUDGR}"
mensaje_rot13 = "SYNT{FRPERG_SYNT_EBBG13}"

# Descifrar
print("César descifrado:", cesar_descifrar(mensaje_cesar, 3))
print("ROT13 descifrado:", ROT13_descifrar(mensaje_rot13))
```

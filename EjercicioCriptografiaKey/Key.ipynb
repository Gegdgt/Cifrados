## Realizar la creación de un script que permita generar llaves dinámicas (utilizar ASCII)

```python
# Convierte un texto a un número semilla
def texto_a_semilla(txt):
    semilla = 0
    i = 0
    while i < len(txt):
        semilla = semilla * 131 + ord(txt[i])
        semilla = semilla % 2147483647
        i += 1
    if semilla == 0:
        semilla = 123456789
    return semilla

# Generador pseudoaleatorio simple
def siguiente_numero(estado):
    estado = (estado * 1103515245 + 12345) % 2147483647
    return estado

# Pasa un número a ASCII imprimible
def numero_a_ascii_imprimible(num):
    base = 33
    rango = 94
    return chr(base + (num % rango))

# Genera una llave ASCII de longitud n usando una semilla
def generar_llave(semilla_txt, n):
    estado = texto_a_semilla(semilla_txt)
    llave = ""
    i = 0
    while i < n:
        estado = siguiente_numero(estado)
        llave += numero_a_ascii_imprimible(estado)
        i += 1
    return llave

semilla_usuario = input("Semilla (texto o numero): ")
tam = input("Longitud de la llave (numero): ")

n = int(tam)

print("Llave generada:")
print(generar_llave(semilla_usuario, n))
```

## Realizar la creación de un script que generar un nuevo cypher en ASCII con una llave k de tamaño fijo 

```python
# Ajusta la llave a tamaño fijo k (repite o recorta)
def ajustar_llave(llave, k):
    if len(llave) == 0:
        return "A" * k

    nueva = ""
    i = 0
    while i < k:
        nueva += llave[i % len(llave)]
        i += 1
    return nueva

# Cifra con XOR usando ASCII
def cifrar_xor_ascii(mensaje, llave_fija):
    cifrado = []
    i = 0
    while i < len(mensaje):
        m = ord(mensaje[i])
        k = ord(llave_fija[i % len(llave_fija)])
        c = m ^ k
        cifrado.append(c)
        i += 1
    return cifrado

# Convierte números a ASCII imprimible (solo para mostrar)
def a_imprimible(lista):
    txt = ""
    i = 0
    while i < len(lista):
        txt += chr(33 + (lista[i] % 94))
        i += 1
    return txt

mensaje = input("Mensaje: ")
llave = input("Llave (texto): ")
k_txt = input("Tamaño fijo k (numero): ")

k = int(k_txt)

llave_fija = ajustar_llave(llave, k)
cifrado = cifrar_xor_ascii(mensaje, llave_fija)

print("Llave fija usada:")
print(llave_fija)

print("Cifrado (numeros ASCII):")
print(cifrado)

print("Cifrado (ASCII imprimible):")
print(a_imprimible(cifrado))
```

## Realizar la creación de un script que generar un nuevo cypher en ASCII con una llave k de tamaño dinámico


```python
# Convierte texto a lista ASCII
def texto_a_ascii(txt):
    lista = []
    i = 0
    while i < len(txt):
        lista.append(ord(txt[i]))
        i += 1
    return lista

# Pasa un número a ASCII imprimible
def a_imprimible_num(num):
    return chr(33 + (num % 94))

# Genera una llave dinámica de tamaño n desde una llave inicial
def generar_llave_dinamica(llave_inicial, n):
    if len(llave_inicial) == 0:
        llave_inicial = "A"

    base = texto_a_ascii(llave_inicial)
    salida = ""
    i = 0

    while i < n:
        if i < len(base):
            salida += a_imprimible_num(base[i])
        else:
            val = (ord(salida[i - 1]) + ord(salida[i - len(base)])) % 128
            salida += a_imprimible_num(val)
        i += 1

    return salida

# Cifra con XOR usando la llave dinámica
def cifrar_xor_ascii(mensaje, llave_dinamica):
    cifrado = []
    i = 0
    while i < len(mensaje):
        m = ord(mensaje[i])
        k = ord(llave_dinamica[i])
        c = m ^ k
        cifrado.append(c)
        i += 1
    return cifrado

# Convierte números a ASCII imprimible (solo para mostrar)
def lista_a_imprimible(lista):
    txt = ""
    i = 0
    while i < len(lista):
        txt += chr(33 + (lista[i] % 94))
        i += 1
    return txt

mensaje = input("Mensaje: ")
llave_ini = input("Llave inicial: ")

llave_din = generar_llave_dinamica(llave_ini, len(mensaje))
cifrado = cifrar_xor_ascii(mensaje, llave_din)

print("Llave dinamica usada:")
print(llave_din)

print("Cifrado (numeros ASCII):")
print(cifrado)

print("Cifrado (ASCII imprimible):")
print(lista_a_imprimible(cifrado))
```

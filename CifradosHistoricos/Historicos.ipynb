1. Cifrado César

```python
def cesar_cifrar(msj, d):
    # Recorre cada caracter y desplaza solo letras A-Z / a-z
    res = ""
    d = d % 26  # evita desplazamientos enormes

    for ch in msj:
        o = ord(ch)

        # Mayúsculas
        if o >= ord('A') and o <= ord('Z'):
            base = ord('A')
            pos = o - base
            pos2 = (pos + d) % 26
            res += chr(base + pos2)

        # Minúsculas
        elif o >= ord('a') and o <= ord('z'):
            base = ord('a')
            pos = o - base
            pos2 = (pos + d) % 26
            res += chr(base + pos2)

        # Otros caracteres no cambian
        else:
            res += ch

    return res


def cesar_descifrar(msj, d):
    # Descifrar es desplazar al lado contrario
    return cesar_cifrar(msj, -d)
```

2. Cifrado ROT13

```python
def rot13(msj):
    # ROT13 rota 13 posiciones en letras; dos veces regresa al original
    def apl(ch):
        o = ord(ch)

        if o >= ord('A') and o <= ord('Z'):
            base = ord('A')
            pos = o - base
            return chr(base + ((pos + 13) % 26))

        if o >= ord('a') and o <= ord('z'):
            base = ord('a')
            pos = o - base
            return chr(base + ((pos + 13) % 26))

        return ch  # no letras

    res = ""
    for ch in msj:
        res += apl(ch)
    return res
```

3. Cifrado Vigenere

```python
def vigenere_cifrar(msj, cla):
    # Usa la clave para desplazar letras, ignorar otros
    res = ""
    j = 0  # índice de clave

    for ch in msj:
        o = ord(ch)

        # calculamos el desplazamiento solo si el char es letra
        if (o >= ord('A') and o <= ord('Z')) or (o >= ord('a') and o <= ord('z')):
            # busca una letra válida de la clave
            while True:
                if len(cla) == 0:
                    k = 0
                    break
                kc = cla[j % len(cla)]
                ko = ord(kc)
                if ko >= ord('A') and ko <= ord('Z'):
                    k = ko - ord('A')
                    break
                if ko >= ord('a') and ko <= ord('z'):
                    k = ko - ord('a')
                    break
                j += 1  # si en la clave hay un símbolo raro, lo saltamos

            # cifrar según sea mayúscula o minúscula
            if o >= ord('A') and o <= ord('Z'):
                base = ord('A')
                pos = o - base
                res += chr(base + ((pos + k) % 26))
            else:
                base = ord('a')
                pos = o - base
                res += chr(base + ((pos + k) % 26))

            j += 1
        else:
            res += ch

    return res


def vigenere_descifrar(msj, cla):
    # Igual que cifrar, pero resta el desplazamiento
    res = ""
    j = 0

    for ch in msj:
        o = ord(ch)

        if (o >= ord('A') and o <= ord('Z')) or (o >= ord('a') and o <= ord('z')):
            while True:
                if len(cla) == 0:
                    k = 0
                    break
                kc = cla[j % len(cla)]
                ko = ord(kc)
                if ko >= ord('A') and ko <= ord('Z'):
                    k = ko - ord('A')
                    break
                if ko >= ord('a') and ko <= ord('z'):
                    k = ko - ord('a')
                    break
                j += 1

            if o >= ord('A') and o <= ord('Z'):
                base = ord('A')
                pos = o - base
                res += chr(base + ((pos - k) % 26))
            else:
                base = ord('a')
                pos = o - base
                res += chr(base + ((pos - k) % 26))

            j += 1
        else:
            res += ch

    return res
```

4. Analisis de Frecuencias

```python
def analisis_frecuencia(msj):
    # Cuenta cuántas veces aparece cada caracter
    fre = {}
    tot = 0

    for ch in msj:
        tot += 1
        if ch in fre:
            fre[ch] = fre[ch] + 1
        else:
            fre[ch] = 1

    # ordenar por frecuencia (de mayor a menor)
    it = list(fre.items())
    it.sort(key=lambda x: x[1], reverse=True)

    if tot == 0:
        print("Mensaje vacio")
        return

    # caracter más usado
    mas = it[0][0]
    cant = it[0][1]
    porc = (cant * 100.0) / tot

    # manejo visual de caracteres especiales
    if mas == " ":
        m = "[espacio]"
    elif mas == "\n":
        m = "[salto de linea]"
    elif mas == "\t":
        m = "[tab]"
    else:
        m = mas

    print("El caracter mas usado es:", m)
    print("")

    # detalle de cada caracter
    for par in it:
        c = par[0]
        n = par[1]
        p = (n * 100.0) / tot

        if c == " ":
            cs = "[espacio]"
        elif c == "\n":
            cs = "[salto de linea]"
        elif c == "\t":
            cs = "[tab]"
        else:
            cs = c

        print("El caracter", cs, "se utilizo", n, "veces, lo cual representa el", round(p, 2), "% de la palabra")
```

```python
msj = input("Mensaje a Cifrar: ")
d = int(input("Desplazamiento Cesar (numero): "))
cla = input("Clave Vigenere (solo letras recomendado): ")

print("Cesar cifrado:")
print(cesar_cifrar(msj, d))
print("Cesar descifrado:")
print(cesar_descifrar(cesar_cifrar(msj, d), d))

print("ROT13:")
print(rot13(msj))
print("ROT13 descifrado:")
print(rot13(rot13(msj)))

print("Vigenere cifrado:")
print(vigenere_cifrar(msj, cla))
print("Vigenere descifrado:")
print(vigenere_descifrar(vigenere_cifrar(msj, cla), cla))

print("\n\nFrecuencias del mensaje original:")
analisis_frecuencia(msj)
```

```python
abc_may = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
abc_min = "abcdefghijklmnopqrstuvwxyz"
```

1. Cifrado César

```python
def cesar_cifrar(msj, d):
    # Diccionarios de sustitucion
    may = {}
    min = {}
    
    d = d % 26  # ajusta el desplazamiento

    # construir diccionarios
    for i in range(26):
        may[abc_may[i]] = abc_may[(i + d) % 26]
        min[abc_min[i]] = abc_min[(i + d) % 26]

    res = ""

    # cifrar caracter por caracter
    for ch in msj:
        if ch in may:
            res += may[ch]
        elif ch in min:
            res += min[ch]
        else:
            res += ch

    return res
```

```python
def cesar_descifrar(msj, d):
    # Diccionarios de sustitucion inversa
    may = {}
    min = {}

    d = d % 26

    # construir diccionarios inversos
    for i in range(26):
        may[abc_may[(i + d) % 26]] = abc_may[i]
        min[abc_min[(i + d) % 26]] = abc_min[i]

    res = ""

    # descifrar caracter por caracter
    for ch in msj:
        if ch in may:
            res += may[ch]
        elif ch in min:
            res += min[ch]
        else:
            res += ch

    return res
```

2. Cifrado ROT13

```python
def rot13(msj):
    return cesar_cifrar(msj, 13)
```

3. Cifrado Vigenere

```python
def vigenere_cifrar(msj, cla):
    # Usa la clave para desplazar letras, ignorar otros
    res = ""
    j = 0  # índice de clave

    for ch in msj:
        o = ord(ch)

        # calculamos el desplazamiento solo si el char es letra
        if (o >= ord('A') and o <= ord('Z')) or (o >= ord('a') and o <= ord('z')):
            # busca una letra válida de la clave
            while True:
                if len(cla) == 0:
                    k = 0
                    break
                kc = cla[j % len(cla)]
                ko = ord(kc)
                if ko >= ord('A') and ko <= ord('Z'):
                    k = ko - ord('A')
                    break
                if ko >= ord('a') and ko <= ord('z'):
                    k = ko - ord('a')
                    break
                j += 1  # si en la clave hay un símbolo raro, lo saltamos

            # cifrar según sea mayúscula o minúscula
            if o >= ord('A') and o <= ord('Z'):
                base = ord('A')
                pos = o - base
                res += chr(base + ((pos + k) % 26))
            else:
                base = ord('a')
                pos = o - base
                res += chr(base + ((pos + k) % 26))

            j += 1
        else:
            res += ch

    return res
```

```python
def vigenere_descifrar(msj, cla):
    # Igual que cifrar, pero resta el desplazamiento
    res = ""
    j = 0

    for ch in msj:
        o = ord(ch)

        if (o >= ord('A') and o <= ord('Z')) or (o >= ord('a') and o <= ord('z')):
            while True:
                if len(cla) == 0:
                    k = 0
                    break
                kc = cla[j % len(cla)]
                ko = ord(kc)
                if ko >= ord('A') and ko <= ord('Z'):
                    k = ko - ord('A')
                    break
                if ko >= ord('a') and ko <= ord('z'):
                    k = ko - ord('a')
                    break
                j += 1

            if o >= ord('A') and o <= ord('Z'):
                base = ord('A')
                pos = o - base
                res += chr(base + ((pos - k) % 26))
            else:
                base = ord('a')
                pos = o - base
                res += chr(base + ((pos - k) % 26))

            j += 1
        else:
            res += ch

    return res
```

4. Analisis de Frecuencias

```python
def analisis_frecuencia(msj):
    # Cuenta solo letras y calcula su probabilidad
    fre = {}
    tot = 0

    for ch in msj:
        o = ord(ch)

        # convertir mayúsculas a minúsculas
        if o >= ord('A') and o <= ord('Z'):
            ch = chr(o + 32)

        # solo letras a-z
        if ch >= 'a' and ch <= 'z':
            tot += 1
            if ch in fre:
                fre[ch] = fre[ch] + 1
            else:
                fre[ch] = 1

    if tot == 0:
        print("No hay letras en el mensaje")
        return

    # ordenar por frecuencia
    it = list(fre.items())
    it.sort(key=lambda x: x[1], reverse=True)

    # letra más usada
    mas = it[0][0]
    cant = it[0][1]
    porc = (cant * 100.0) / tot

    print("La letra mas usada es:", mas)
    print("")

    # mostrar probabilidad de cada letra
    for par in it:
        c = par[0]
        n = par[1]
        p = (n * 100.0) / tot
        print("", c, ":", n, "veces --> ", round(p, 2), "%")
```

```python
def comparar_frecuencias(msj):
    # tabla real del español (aproximada)
    real = {
        'a': 12.53, 'b': 1.42, 'c': 4.68, 'd': 5.86, 'e': 13.68,
        'f': 0.69, 'g': 1.01, 'h': 0.70, 'i': 6.25, 'j': 0.44,
        'k': 0.02, 'l': 4.97, 'm': 3.15, 'n': 6.71, 'o': 8.68,
        'p': 2.51, 'q': 0.88, 'r': 6.87, 's': 7.98, 't': 4.63,
        'u': 3.93, 'v': 0.90, 'w': 0.01, 'x': 0.22, 'y': 0.90,
        'z': 0.52, 'ñ': 0.31
    }

    fre = {}
    tot = 0

    for ch in msj:
        o = ord(ch)

        if o >= ord('A') and o <= ord('Z'):
            ch = chr(o + 32)

        if (ch >= 'a' and ch <= 'z') or ch == 'ñ':
            tot += 1
            if ch in fre:
                fre[ch] = fre[ch] + 1
            else:
                fre[ch] = 1

    if tot == 0:
        print("No hay letras para analizar")
        return

    print("Comparacion de frecuencias e incertidumbre \n")

    for l in real:
        obs = 0
        if l in fre:
            obs = (fre[l] * 100.0) / tot

        esp = real[l]
        inc = abs(esp - obs)

        print("", l, "--> Teorico:", esp, "%, Practico:", round(obs, 2), "%, Incertidumbre:", round(inc, 2), "%")
```

```python
msj = input("Mensaje a Cifrar: ")
d = int(input("Desplazamiento Cesar (numero): "))
cla = input("Clave Vigenere (solo letras recomendado): ")

print("\nPalabra Original:", msj)

print("\nCesar cifrado:")
print(cesar_cifrar(msj, d))
print("\nCesar descifrado:")
print(cesar_descifrar(cesar_cifrar(msj, d), d))

print("\nROT13:")
print(rot13(msj))
print("\nROT13 descifrado:")
print(cesar_descifrar(rot13(msj), 13))

print("\nVigenere cifrado:")
print(vigenere_cifrar(msj, cla))
print("\nVigenere descifrado:")
print(vigenere_descifrar(vigenere_cifrar(msj, cla), cla))

print("\n\nFrecuencias del mensaje original:")
analisis_frecuencia(msj)

print("\n\nComparacion de frecuencias del mensaje original con el español:")
comparar_frecuencias(msj)
```
